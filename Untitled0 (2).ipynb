```python
from tqdm import tqdm


projected_corpus = []
errors = 0


for item in tqdm(alignment_data, desc="Projecting POS tags"):
    hindi_tokens = item['tokens_hindi']
    kumaoni_tokens = item['tokens_kumaoni']
    alignments = item['alignments']


    if len(alignments) == 0:
        continue


    hindi_sentence = ' '.join(hindi_tokens)
    try:
        doc = nlp(hindi_sentence)
    except Exception as e:
        errors += 1
        continue


    hindi_pos = []
    for sent in doc.sentences:
        for word in sent.words:
            hindi_pos.append(word.pos)


    if len(hindi_pos) != len(hindi_tokens):
        errors += 1
        continue


    kumaoni_pos = ['UNK'] * len(kumaoni_tokens)


    for align in alignments:
        h_idx = align['hindi_idx']
        k_idx = align['kumaoni_idx']


        if h_idx < len(hindi_pos) and k_idx < len(kumaoni_tokens):
            kumaoni_pos[k_idx] = hindi_pos[h_idx]


    projected_corpus.append({
        'hindi_tokens': hindi_tokens,
        'kumaoni_tokens': kumaoni_tokens,
        'hindi_pos': hindi_pos,
        'kumaoni_pos': kumaoni_pos
    })

print(f"\nProjection complete!")
print(f"Successfully projected: {len(projected_corpus)} sentences")
print(f"Errors/skipped: {errors}")
```

```python
import json
import torch
import torch.nn as nn
from collections import Counter


with open('/content/kumaoni_training_data.json', 'r', encoding='utf-8') as f:
    training_data = json.load(f)

print(f"Sentences loaded: {len(training_data)}")
print(f"Total tokens: {sum(len(e['words']) for e in training_data)}")


word_counter = Counter()
char_counter = Counter()
tag_set = set()

for entry in training_data:
    for word, tag in zip(entry['words'], entry['tags']):
        word_counter[word] += 1
        tag_set.add(tag)

        for ch in word:
            char_counter[ch] += 1


PAD_WORD = '<PAD>'
UNK_WORD = '<UNK>'
PAD_CHAR = '<PAD>'
PAD_TAG = '<PAD>'


word2idx = {PAD_WORD: 0, UNK_WORD: 1}
for word, _ in word_counter.most_common():
    word2idx[word] = len(word2idx)


char2idx = {PAD_CHAR: 0}
for ch, _ in char_counter.most_common():
    char2idx[ch] = len(char2idx)

tag2idx = {PAD_TAG: 0}
for tag in sorted(tag_set):
    tag2idx[tag] = len(tag2idx)

idx2tag = {i: t for t, i in tag2idx.items()}

print(f"\nWord vocab size: {len(word2idx)}")
print(f"Char vocab size: {len(char2idx)}")
print(f"Tag vocab size: {len(tag2idx)}")
print(f"\nTags: {sorted(tag_set)}")
```

```python
from torch.utils.data import Dataset, DataLoader

class KumaoniPOSDataset(Dataset):
    def __init__(self, data, word2idx, char2idx, tag2idx, max_word_len=15):
        self.data = data
        self.word2idx = word2idx
        self.char2idx = char2idx
        self.tag2idx = tag2idx
        self.max_word_len = max_word_len

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        entry = self.data[idx]
        words = entry['words']
        tags = entry['tags']


        word_ids = [self.word2idx.get(w, self.word2idx['<UNK>']) for w in words]
        tag_ids = [self.tag2idx[t] for t in tags]


        char_ids = []
        for w in words:
            chars = [self.char2idx.get(c, 0) for c in w[:self.max_word_len]]

            chars = chars + [0] * (self.max_word_len - len(chars))
            char_ids.append(chars)

        return {
            'word_ids': torch.tensor(word_ids, dtype=torch.long),
            'char_ids': torch.tensor(char_ids, dtype=torch.long),
            'tag_ids': torch.tensor(tag_ids, dtype=torch.long)
        }

def collate_fn(batch):

    max_len = max(len(item['word_ids']) for item in batch)

    word_ids_padded = []
    char_ids_padded = []
    tag_ids_padded = []
    mask = []

    for item in batch:
        length = len(item['word_ids'])
        pad_len = max_len - length
        word_ids_padded.append(torch.cat([item['word_ids'], torch.zeros(pad_len, dtype=torch.long)]))
        char_ids_padded.append(torch.cat([item['char_ids'], torch.zeros(pad_len, item['char_ids'].size(1), dtype=torch.long)]))
        tag_ids_padded.append(torch.cat([item['tag_ids'], torch.zeros(pad_len, dtype=torch.long)]))
        mask.append(torch.cat([torch.ones(length), torch.zeros(pad_len)]))

    return {
        'word_ids': torch.stack(word_ids_padded),
        'char_ids': torch.stack(char_ids_padded),
        'tag_ids': torch.stack(tag_ids_padded),
        'mask': torch.stack(mask).bool()
    }


import random
random.seed(42)
random.shuffle(training_data)
split = int(0.9 * len(training_data))
train_data = training_data[:split]
dev_data = training_data[split:]

train_dataset = KumaoniPOSDataset(train_data, word2idx, char2idx, tag2idx)
dev_dataset = KumaoniPOSDataset(dev_data, word2idx, char2idx, tag2idx)

train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True, collate_fn=collate_fn)
dev_loader = DataLoader(dev_dataset, batch_size=16, shuffle=False, collate_fn=collate_fn)

print(f"Train sentences: {len(train_dataset)}")
print(f"Dev sentences: {len(dev_dataset)}")
print(f"Train batches: {len(train_loader)}")
print(f"Dev batches: {len(dev_loader)}")
```

```python



class CRF(nn.Module):
    def __init__(self, num_tags):
        super(CRF, self).__init__()
        self.num_tags = num_tags
        self.transitions = nn.Parameter(torch.randn(num_tags, num_tags))
        self.start_transitions = nn.Parameter(torch.randn(num_tags))
        self.end_transitions = nn.Parameter(torch.randn(num_tags))

    def forward(self, emissions, tags, mask):
        log_likelihood = (
            self._compute_log_likelihood(emissions, tags, mask)
            - self._compute_normalizer(emissions, mask)
        )
        return -log_likelihood.mean()

    def _compute_log_likelihood(self, emissions, tags, mask):
        batch_size, seq_len, num_tags = emissions.shape

        score = (
            self.start_transitions[tags[:, 0]]
            + emissions[:, 0].gather(1, tags[:, :1]).squeeze(1)
        )

        for i in range(1, seq_len):
            score += self.transitions[tags[:, i - 1], tags[:, i]] * mask[:, i]
            score += emissions[:, i].gather(1, tags[:, i:i + 1]).squeeze(1) * mask[:, i]

        lengths = mask.sum(dim=1).long() - 1
        score += self.end_transitions[tags.gather(1, lengths.unsqueeze(1)).squeeze(1)]
        return score

    def _compute_normalizer(self, emissions, mask):
        batch_size, seq_len, num_tags = emissions.shape

        alpha = self.start_transitions + emissions[:, 0]          # [B, T]
        for i in range(1, seq_len):
            # scores[b, prev, cur] = alpha[b, prev] + trans[prev, cur] + emit[b, cur]
            scores = (
                alpha.unsqueeze(2)                                # [B, T, 1] — prev
                + self.transitions.unsqueeze(0)                   # [1, T, T]
                + emissions[:, i].unsqueeze(1)                    # [B, 1, T] — cur
            )
            new_alpha = torch.logsumexp(scores, dim=1)            # [B, T] reduce over prev
            alpha = torch.where(mask[:, i].unsqueeze(1), new_alpha, alpha)

        alpha = alpha + self.end_transitions
        return torch.logsumexp(alpha, dim=1)                      # [B]

    def decode(self, emissions, mask):
        batch_size, seq_len, num_tags = emissions.shape

        score = self.start_transitions + emissions[:, 0]
        history = []

        for i in range(1, seq_len):
            broadcast_score = score.unsqueeze(2)
            broadcast_emission = emissions[:, i].unsqueeze(1)
            next_score = broadcast_score + self.transitions + broadcast_emission

            best_score, best_tag = next_score.max(dim=1)

            score = torch.where(mask[:, i].unsqueeze(1), best_score, score)
            history.append(best_tag)

        score += self.end_transitions
        best_last_tag = score.argmax(dim=1)

        best_tags_list = []
        for b in range(batch_size):
            length = int(mask[b].sum().item())

            best_tags = [best_last_tag[b].item()]
            for t in range(length - 2, -1, -1):
                prev_tag = best_tags[-1]
                best_tags.append(history[t][b, prev_tag].item())
            best_tags.reverse()
            best_tags_list.append(best_tags)

        return best_tags_list


print("CRF defined.")
```

```python
class BiLSTMCRF(nn.Module):
    def __init__(self, word_vocab_size, char_vocab_size, num_tags,
                 word_emb_dim=100, char_emb_dim=30, char_hidden_dim=50,
                 word_hidden_dim=128, dropout=0.3, max_word_len=15):
        super(BiLSTMCRF, self).__init__()

        self.max_word_len = max_word_len


        self.word_emb = nn.Embedding(word_vocab_size, word_emb_dim, padding_idx=0)


        self.char_emb = nn.Embedding(char_vocab_size, char_emb_dim, padding_idx=0)
        self.char_lstm = nn.LSTM(char_emb_dim, char_hidden_dim,
                                 bidirectional=True, batch_first=True)


        self.word_lstm = nn.LSTM(
            input_size=word_emb_dim + 2 * char_hidden_dim,
            hidden_size=word_hidden_dim,
            num_layers=1,
            bidirectional=True,
            batch_first=True
        )

        self.dropout = nn.Dropout(dropout)


        self.hidden2tag = nn.Linear(2 * word_hidden_dim, num_tags)


        self.crf = CRF(num_tags)

    def _char_repr(self, char_ids):

        batch_size, seq_len, max_word_len = char_ids.shape

        char_ids_flat = char_ids.view(-1, max_word_len)
        char_embs = self.char_emb(char_ids_flat)
        _, (hidden, _) = self.char_lstm(char_embs)

        hidden = torch.cat([hidden[0], hidden[1]], dim=1)
        return hidden.view(batch_size, seq_len, -1)

    def forward(self, word_ids, char_ids, mask, tags=None):

        word_embs = self.word_emb(word_ids)


        char_repr = self._char_repr(char_ids)


        combined = torch.cat([word_embs, char_repr], dim=2)
        combined = self.dropout(combined)


        lstm_out, _ = self.word_lstm(combined)  # [batch, seq_len, 2*word_hidden_dim]
        lstm_out = self.dropout(lstm_out)


        emissions = self.hidden2tag(lstm_out)  # [batch, seq_len, num_tags]

        if tags is not None:
            # Training: compute loss
            loss = self.crf(emissions, tags, mask)
            return loss
        else:
            # Inference: Viterbi decode
            return self.crf.decode(emissions, mask)


model = BiLSTMCRF(
    word_vocab_size=len(word2idx),
    char_vocab_size=len(char2idx),
    num_tags=len(tag2idx)
)


device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = model.to(device)


total_params = sum(p.numel() for p in model.parameters())
print(f"Device: {device}")
print(f"Total model parameters: {total_params:,}")
```

```python
import torch.optim as optim

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = model.to(device)

optimizer = optim.Adam(model.parameters(), lr=1e-3, weight_decay=1e-5)

print(f"Device: {device}")
print(f"Params: {sum(p.numel() for p in model.parameters()):,}")
```

```python
import time

def train_epoch(model, loader, optimizer, device):
    model.train()
    total_loss = 0.0
    n_batches = 0
    for batch in loader:
        word_ids = batch['word_ids'].to(device)
        char_ids = batch['char_ids'].to(device)
        tag_ids  = batch['tag_ids'].to(device)
        mask     = batch['mask'].to(device)

        optimizer.zero_grad()
        loss = model(word_ids, char_ids, mask, tags=tag_ids)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=5.0)
        optimizer.step()

        total_loss += loss.item()
        n_batches += 1
    return total_loss / max(n_batches, 1)


def evaluate_loss(model, loader, device):
    model.eval()
    total_loss = 0.0
    n_batches = 0
    with torch.no_grad():
        for batch in loader:
            word_ids = batch['word_ids'].to(device)
            char_ids = batch['char_ids'].to(device)
            tag_ids  = batch['tag_ids'].to(device)
            mask     = batch['mask'].to(device)
            loss = model(word_ids, char_ids, mask, tags=tag_ids)
            total_loss += loss.item()
            n_batches += 1
    return total_loss / max(n_batches, 1)


EPOCHS = 20
best_dev = float('inf')
patience = 5
bad_epochs = 0

for epoch in range(1, EPOCHS + 1):
    t0 = time.time()
    train_loss = train_epoch(model, train_loader, optimizer, device)
    dev_loss   = evaluate_loss(model, dev_loader, device)
    dt = time.time() - t0

    print(f"Epoch {epoch:02d} | train {train_loss:.4f} | dev {dev_loss:.4f} | {dt:.1f}s")

    if dev_loss < best_dev:
        best_dev = dev_loss
        bad_epochs = 0
        torch.save(model.state_dict(), 'best_model.pt')
    else:
        bad_epochs += 1
        if bad_epochs >= patience:
            print(f"Early stopping at epoch {epoch} (best dev {best_dev:.4f})")
            break

print(f"\nBest dev loss: {best_dev:.4f}")
```

```python
from sklearn.metrics import classification_report, accuracy_score, f1_score

def evaluate_predictions(model, loader, device, idx2tag):
    model.eval()
    all_true, all_pred = [], []
    with torch.no_grad():
        for batch in loader:
            word_ids = batch['word_ids'].to(device)
            char_ids = batch['char_ids'].to(device)
            tag_ids  = batch['tag_ids'].to(device)
            mask     = batch['mask'].to(device)

            preds = model(word_ids, char_ids, mask)

            for b in range(word_ids.size(0)):
                length = int(mask[b].sum().item())
                true_seq = tag_ids[b, :length].tolist()
                pred_seq = preds[b]
                all_true.extend(true_seq)
                all_pred.extend(pred_seq)

    y_true = [idx2tag[t] for t in all_true]
    y_pred = [idx2tag[t] for t in all_pred]
    return y_true, y_pred



model.load_state_dict(torch.load('best_model.pt', map_location=device))

y_true, y_pred = evaluate_predictions(model, dev_loader, device, idx2tag)

print(f"Token accuracy: {accuracy_score(y_true, y_pred):.4f}")
print(f"Macro F1:       {f1_score(y_true, y_pred, average='macro'):.4f}")
print(f"Weighted F1:    {f1_score(y_true, y_pred, average='weighted'):.4f}")
print()
print(classification_report(y_true, y_pred, zero_division=0))
```

```python
def tag_sentence(text, model, word2idx, char2idx, idx2tag, max_word_len=15, device='cpu'):
    model.eval()
    words = text.split()
    word_ids = [word2idx.get(w, word2idx['<UNK>']) for w in words]
    char_ids = []
    for w in words:
        chars = [char2idx.get(c, 0) for c in w[:max_word_len]]
        chars += [0] * (max_word_len - len(chars))
        char_ids.append(chars)

    word_t = torch.tensor([word_ids], dtype=torch.long).to(device)
    char_t = torch.tensor([char_ids], dtype=torch.long).to(device)
    mask_t = torch.ones(1, len(words), dtype=torch.bool).to(device)

    with torch.no_grad():
        preds = model(word_t, char_t, mask_t)

    return list(zip(words, [idx2tag[t] for t in preds[0]]))


sample = "यो लौंड रा्णिखेत बटि गा्ड़ि लिबेर या्ं पैलि पैलि ऊंणौ ।"
for w, t in tag_sentence(sample, model, word2idx, char2idx, idx2tag, device=device):
    print(f"{w:20} {t}")
```
